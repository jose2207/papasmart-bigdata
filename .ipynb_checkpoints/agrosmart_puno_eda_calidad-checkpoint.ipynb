```
from pyspark.sql import SparkSession
from pyspark.sql.types import StructType, StructField, StringType, IntegerType, DoubleType, TimestampType, BooleanType
from pyspark.sql import functions as F

spark = (
    SparkSession.builder
    .appName("agrosmart-puno-eda")
    .master("local[*]")
    .config("spark.ui.port", "4040")
    .getOrCreate()
)

spark
```

## 1. Carga con esquema explícito

Se define `StructType` para las tres fuentes en vez de dejar que Spark infiera el tipo — así se detectan desde la lectura columnas mal tipadas (por ejemplo, fechas leídas como texto).

```
RUTA_CSV_LECTURAS = "file:///opt/solanumtuberosumdata/DATA/lecturas_humedad_suelo.csv"
RUTA_CSV_PERFIL = "file:///opt/solanumtuberosumdata/DATA/perfil_diario_riego_clima.csv"
RUTA_CSV_METADATA = "file:///opt/solanumtuberosumdata/DATA/sensores_metadata.csv"

schema_lecturas = StructType([
    StructField("reading_id", IntegerType(), True),
    StructField("sensor_id", StringType(), True),
    StructField("timestamp", TimestampType(), True),
    StructField("humedad_suelo_pct", DoubleType(), True),
    StructField("temp_suelo_c", DoubleType(), True),
    StructField("temp_aire_c", DoubleType(), True),
    StructField("humedad_aire_pct", DoubleType(), True),
    StructField("lluvia_mm_ultima_hora", DoubleType(), True),
    StructField("bateria_v", DoubleType(), True),
    StructField("señal_dbm", IntegerType(), True),
    StructField("estado_sensor", StringType(), True),
])

schema_perfil = StructType([
    StructField("sensor_id", StringType(), True),
    StructField("fecha", StringType(), True),  # se castea a date abajo
    StructField("etapa_cultivo", StringType(), True),
    StructField("dias_desde_siembra", IntegerType(), True),
    StructField("humedad_prom_pct", DoubleType(), True),
    StructField("humedad_min_pct", DoubleType(), True),
    StructField("humedad_max_pct", DoubleType(), True),
    StructField("lluvia_mm", DoubleType(), True),
    StructField("riego_aplicado", BooleanType(), True),
    StructField("volumen_riego_l", DoubleType(), True),
    StructField("evapotranspiracion_mm", DoubleType(), True),
    StructField("alerta", StringType(), True),
])

schema_metadata = StructType([
    StructField("sensor_id", StringType(), True),
    StructField("nombre_estacion", StringType(), True),
    StructField("region", StringType(), True),
    StructField("latitud", DoubleType(), True),
    StructField("longitud", DoubleType(), True),
    StructField("altitud_m", IntegerType(), True),
    StructField("cultivo", StringType(), True),
    StructField("tipo_suelo", StringType(), True),
    StructField("profundidad_sensor_cm", IntegerType(), True),
    StructField("marca_sensor", StringType(), True),
    StructField("sistema_riego", StringType(), True),
    StructField("area_parcela_ha", DoubleType(), True),
    StructField("fecha_instalacion", StringType(), True),
])

df_lecturas = spark.read.csv(RUTA_CSV_LECTURAS, header=True, schema=schema_lecturas)
df_perfil = spark.read.csv(RUTA_CSV_PERFIL, header=True, schema=schema_perfil).withColumn("fecha", F.to_date("fecha"))
df_metadata = spark.read.csv(RUTA_CSV_METADATA, header=True, schema=schema_metadata)

print("lecturas:", df_lecturas.count(), "filas")
print("perfil:", df_perfil.count(), "filas")
print("metadata:", df_metadata.count(), "filas")

df_lecturas.printSchema()
```

## 2. Filtrado al alcance del proyecto: región Puno

El dataset crudo cubre 120 estaciones en 10 regiones y 12 cultivos. AgroSmart Puno delimita su análisis a las **7 estaciones de Puno**, declaradas en el Brief técnico-analítico (sección 2).

```
df_metadata_puno = df_metadata.filter(F.col("region") == "Puno")
sensores_puno = [r["sensor_id"] for r in df_metadata_puno.select("sensor_id").collect()]
print("Estaciones de Puno:", sensores_puno)

df_metadata_puno.select(
    "sensor_id", "nombre_estacion", "cultivo", "altitud_m", "tipo_suelo", "sistema_riego", "area_parcela_ha"
).show(truncate=False)
```

```
df_lecturas_puno = df_lecturas.join(F.broadcast(df_metadata_puno.select("sensor_id")), on="sensor_id", how="inner")
df_perfil_puno = df_perfil.join(F.broadcast(df_metadata_puno.select("sensor_id")), on="sensor_id", how="inner")

print("lecturas Puno:", df_lecturas_puno.count(), "filas")
print("perfil Puno:", df_perfil_puno.count(), "filas")
```

## 3. Diagnóstico de calidad: nulos

Se cuenta explícitamente los nulos por columna en las tres fuentes, ya filtradas a Puno — es la base para decidir, en el pipeline de transformación (S3), si se imputan (`.na.fill()`) o se descartan (`.na.drop()`).

```
def contar_nulos(df_spark, nombre):
    total = df_spark.count()
    conteo = df_spark.select([
        F.count(F.when(F.col(c).isNull(), c)).alias(c) for c in df_spark.columns
    ]).toPandas().T
    conteo.columns = ["nulos"]
    conteo["pct_nulos"] = (conteo["nulos"] / total * 100).round(2)
    conteo["fuente"] = nombre
    return conteo

nulos_lecturas = contar_nulos(df_lecturas_puno, "lecturas_humedad_suelo")
nulos_perfil = contar_nulos(df_perfil_puno, "perfil_diario_riego_clima")

nulos_lecturas
```

```
nulos_perfil
```

## 4. Gráficos de calidad de datos

Se convierte a pandas (`toPandas()`) solo para graficar — el procesamiento y los conteos siguen distribuidos en Spark, esto es exclusivamente para visualización con matplotlib.

```
import matplotlib.pyplot as plt
import pandas as pd

fig, axes = plt.subplots(1, 2, figsize=(13, 5))

nulos_lecturas_plot = nulos_lecturas[nulos_lecturas["nulos"] > 0]
axes[0].bar(nulos_lecturas_plot.index, nulos_lecturas_plot["nulos"], color="#4C72B0")
axes[0].set_title("Nulos por columna — lecturas_humedad_suelo (Puno)")
axes[0].set_ylabel("Cantidad de nulos")
axes[0].tick_params(axis="x", rotation=45)

nulos_perfil_plot = nulos_perfil[nulos_perfil["nulos"] > 0]
axes[1].bar(nulos_perfil_plot.index, nulos_perfil_plot["nulos"], color="#DD8452")
axes[1].set_title("Nulos por columna — perfil_diario_riego_clima (Puno)")
axes[1].set_ylabel("Cantidad de nulos")
axes[1].tick_params(axis="x", rotation=45)

plt.tight_layout()
plt.savefig("nulos_por_columna.png", dpi=110)
plt.show()
```

```
# Duplicados: filas repetidas por sensor_id + timestamp (lecturas) y sensor_id + fecha (perfil)
dup_lecturas = df_lecturas_puno.groupBy("sensor_id", "timestamp").count().filter(F.col("count") > 1).count()
dup_perfil = df_perfil_puno.groupBy("sensor_id", "fecha").count().filter(F.col("count") > 1).count()

print("Duplicados en lecturas (mismo sensor+hora):", dup_lecturas)
print("Duplicados en perfil (mismo sensor+día):", dup_perfil)
```

```
# Distribución de humedad de suelo (lecturas en vivo) por estación de Puno
pdf_humedad = df_lecturas_puno.select("sensor_id", "humedad_suelo_pct").dropna().toPandas()

fig, ax = plt.subplots(figsize=(9, 5))
pdf_humedad.boxplot(column="humedad_suelo_pct", by="sensor_id", ax=ax, grid=False)
ax.set_title("Distribución de humedad de suelo por estación — Puno")
ax.set_xlabel("Estación (sensor_id)")
ax.set_ylabel("Humedad de suelo (%)")
plt.suptitle("")
plt.tight_layout()
plt.savefig("boxplot_humedad_por_estacion.png", dpi=110)
plt.show()
```

```
# Humedad promedio diaria por etapa de cultivo (perfil), para dimensionar el modelo de regresión (Jose, U1)
pdf_etapa = (
    df_perfil_puno.groupBy("etapa_cultivo")
    .agg(F.avg("humedad_prom_pct").alias("humedad_prom"), F.count("*").alias("n_dias"))
    .toPandas()
    .sort_values("humedad_prom")
)

fig, ax = plt.subplots(figsize=(8, 5))
ax.barh(pdf_etapa["etapa_cultivo"], pdf_etapa["humedad_prom"], color="#55A868")
ax.set_xlabel("Humedad de suelo promedio (%)")
ax.set_title("Humedad de suelo promedio por etapa de cultivo — Puno")
plt.tight_layout()
plt.savefig("humedad_promedio_por_etapa.png", dpi=110)
plt.show()

pdf_etapa
```

```
# Frecuencia de alertas de saturación por estación (perfil), para dimensionar el modelo de clasificación (Mijael, U1)
pdf_alertas = (
    df_perfil_puno.groupBy("sensor_id")
    .agg(
        F.count("*").alias("dias_totales"),
        F.sum(F.when(F.col("alerta") == "SATURACION_SUELO", 1).otherwise(0)).alias("dias_con_alerta"),
    )
    .withColumn("pct_alerta", F.round(F.col("dias_con_alerta") / F.col("dias_totales") * 100, 2))
    .orderBy(F.desc("pct_alerta"))
    .toPandas()
)

fig, ax = plt.subplots(figsize=(8, 5))
ax.bar(pdf_alertas["sensor_id"], pdf_alertas["pct_alerta"], color="#C44E52")
ax.set_ylabel("% de días con alerta de saturación")
ax.set_title("Frecuencia de alerta de saturación por estación — Puno")
plt.tight_layout()
plt.savefig("pct_alertas_por_estacion.png", dpi=110)
plt.show()

pdf_alertas
```
